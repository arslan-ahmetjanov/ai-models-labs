## 7. Функции

```python
def factorial(n):
    f = 1
    for i in range(1, n+1):
        f *= i
    return f
```

```python
print(factorial(5))
```

## 8. График функции

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.arange(-4,4,0.01)
print(x[0:10])
print(x[-10:])
```

```python
plt.plot(x,x**2)
```

```python
plt.plot(x,x**2)
plt.plot(x,x**3)
```

## 9. Столбцы

```python
values = [12,55,4,32,14]
colors = ['r','g','b','c','m']
plt.bar(range(0,5),values,color =colors)
```

## 10. Круговая диаграмма

```python
values = [12, 55, 4 ,32, 14]
colors = ['r','g','b','c','m']
explode = [0,0,0.2,0,0]
labels = ['One','Two','Three','Four','Five']
plt.title('Diagram')
plt.pie(values, colors = colors, labels = labels, explode= explode)
```

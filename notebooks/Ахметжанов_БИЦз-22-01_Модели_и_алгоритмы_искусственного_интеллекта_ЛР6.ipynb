## Листинг 1. Гистограмма распределения доходов

```python
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
# фиксируем seed: повторный прогон даёт те же числа
np.random.seed(42)
incomes = np.random.normal(27000,15000,10000)
print(incomes[:10])               
```

```python
plt.hist(incomes,50)
plt.show()
```

```python
plt.hist(incomes_1,50)
plt.show()
```

## Листинг 2. Среднее, медиана и мода

```python
import numpy as np
from scipy import stats
```

```python
incomes = np.random.normal(27000,15000,10000)
```

```python
incomes_1 = np.append(incomes, [1000000000,1200000000])
```

```python
print(np.mean(incomes))
print(np.mean(incomes_1))
```

```python
print(np.median(incomes))
print(np.median(incomes_1))
```

```python
print(stats.mode(incomes))
print(stats.mode(incomes_1))
```

## Листинг 3. Меры и диаграммы отклонений

```python
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
incomes = np.random.normal(27000,15000,10000)
```

```python
anomalies = np.random.normal(150000,150,100)
```

```python
incomes_1 = np.concatenate((incomes, anomalies))
```

```python
plt.hist(incomes_1,50)
plt.show()
```

```python
plt.boxplot(incomes)
plt.show()
```

```python
plt.boxplot(incomes_1)
plt.show()
```

```python
print(np.std(incomes))
print(np.std(incomes_1))
```

```python
print(np.var(incomes))
print(np.var(incomes_1))
```

```python
import scipy.stats as sp
```

```python
print(sp.skew(incomes))
print(sp.skew(incomes_1))
```

```python
print(sp.kurtosis(incomes))
print(sp.kurtosis(incomes_1))
```

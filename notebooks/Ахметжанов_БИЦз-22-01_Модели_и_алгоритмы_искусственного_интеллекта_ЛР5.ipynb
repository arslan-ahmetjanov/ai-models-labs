## Листинг 1. Ряды Series

```python
import numpy as np
import pandas as pd
from pandas import Series, DataFrame
```

```python
obj = Series([3,6,9,12])
print(obj)
```

```python
obj.values
```

```python
obj.index
```

```python
obj2 = Series([870,430,300,210,400], index = ['First','Second','Third','Fourth','Fifth'])
print(obj2)
```

```python
obj2['Second']
```

```python
obj2[obj2>400]
```

```python
'Third' in obj2
```

```python
obj2_dictionary = obj2.to_dict()
print(obj2_dictionary)
```

```python
obj3 = Series(obj2_dictionary)
print(obj3)
```

```python
names = ['1st','2nd','3rd','4th','5th']
obj4 = Series(obj2_dictionary, index=names)
print(obj4)
```

```python
names = ['First','Second','Third','4th','5th']
obj4 = Series(obj2_dictionary, index=names)
print(obj4)
```

```python
pd.isnull(obj4)
```

```python
pd.notnull(obj4)
```
